## Question 1 — Basic Python Operations
Accept two numbers and display their sum, difference, product, quotient, and remainder.

## Question 2 — Lists and Tuples
Create a list of 10 numbers; add, remove, and insert elements; find the maximum and minimum; then convert the list into a tuple. Indices used for insertion are **0-based**.

```python
# Question 2: Lists and Tuples
numbers = []
for i in range(10):
    value = int(input(f"Enter number {i + 1}: "))
    numbers.append(value)

print("Original list:", numbers)

# Add an element
add_value = int(input("Enter an element to add: "))
numbers.append(add_value)
print("After adding:", numbers)

# Remove an element by value
remove_value = int(input("Enter an element to remove: "))
if remove_value in numbers:
    numbers.remove(remove_value)
    print("After removing:", numbers)
else:
    print("Element not found; list remains unchanged:", numbers)

# Insert an element at a specified index
insert_value = int(input("Enter an element to insert: "))
position = int(input(f"Enter insertion index (0 to {len(numbers)}): "))
if 0 <= position <= len(numbers):
    numbers.insert(position, insert_value)
    print("After insertion:", numbers)
else:
    print("Invalid index; insertion skipped.")

print("Maximum element:", max(numbers))
print("Minimum element:", min(numbers))

numbers_tuple = tuple(numbers)
print("List converted to tuple:", numbers_tuple)
```

## Question 3 — Strings
Count all characters, count vowels and consonants, reverse the string, and check whether it is a palindrome. For the palindrome check, case, spaces, and punctuation are ignored.

```python
# Question 3: Strings
text = input("Enter a string: ")

print("Number of characters:", len(text))

vowels = 0
consonants = 0
for ch in text.lower():
    if ch.isalpha():
        if ch in "aeiou":
            vowels += 1
        else:
            consonants += 1

print("Number of vowels:", vowels)
print("Number of consonants:", consonants)
print("Reversed string:", text[::-1])

cleaned = "".join(ch.lower() for ch in text if ch.isalnum())
if cleaned == cleaned[::-1]:
    print("The string is a palindrome.")
else:
    print("The string is not a palindrome.")
```

## Question 4 — Sets and Dictionaries
Show union, intersection, both set differences, and symmetric difference. Create a dictionary of five students' marks and print the student with the highest marks.

```python
# Question 4: Sets and Dictionaries
set1 = set(map(int, input("Enter integers for Set 1 (space-separated): ").split()))
set2 = set(map(int, input("Enter integers for Set 2 (space-separated): ").split()))

print("Set 1:", set1)
print("Set 2:", set2)
print("Union:", set1 | set2)
print("Intersection:", set1 & set2)
print("Difference (Set 1 - Set 2):", set1 - set2)
print("Difference (Set 2 - Set 1):", set2 - set1)
print("Symmetric difference:", set1 ^ set2)

students = {}
for i in range(5):
    name = input(f"Enter name of student {i + 1}: ").strip()
    while not name or name in students:
        name = input("Please enter a non-empty, unique name: ").strip()
    marks = float(input(f"Enter marks of {name}: "))
    students[name] = marks

print("Student records:", students)
topper = max(students, key=students.get)
print("Student with the highest marks:", topper)
print("Highest marks:", students[topper])
```

## Question 5 — GPA Calculator
Compute the credit-weighted GPA/SGPA:

\[
\mathrm{GPA} = \frac{\sum(\text{Grade Point} \times \text{Credit})}{\sum\text{Credits}}
\]

Enter numeric **grade points** rather than letter grades. This program assumes grade points range from 0 to 10.

```python
# Question 5: GPA Calculator
num_courses = int(input("Enter number of courses: "))

if num_courses <= 0:
    print("Number of courses must be positive.")
else:
    total_weighted_points = 0.0
    total_credits = 0.0

    for i in range(num_courses):
        print(f"\nCourse {i + 1}")
        grade_point = float(input("Enter grade point (0 to 10): "))
        credit = float(input("Enter credit: "))

        while not (0 <= grade_point <= 10 and credit > 0):
            print("Grade point must be between 0 and 10 and credit must be positive.")
            grade_point = float(input("Enter grade point (0 to 10): "))
            credit = float(input("Enter credit: "))

        total_weighted_points += grade_point * credit
        total_credits += credit

    gpa = total_weighted_points / total_credits
    print(f"\nTotal credits: {total_credits:g}")
    print(f"GPA/SGPA: {gpa:.2f}")
```

## Question 6 — Functions and Iteration
Define separate functions to calculate factorial, generate the Fibonacci series up to **n terms**, and test whether a number is prime.

```python
# Question 6: Functions and Iteration

def factorial(n):
    if n < 0:
        raise ValueError("Factorial is not defined for negative numbers.")
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result


def fibonacci(n):
    series = []
    a, b = 0, 1
    for _ in range(n):
        series.append(a)
        a, b = b, a + b
    return series


def is_prime(n):
    if n < 2:
        return False
    for divisor in range(2, int(n ** 0.5) + 1):
        if n % divisor == 0:
            return False
    return True


factorial_number = int(input("Enter a number to calculate its factorial: "))
try:
    print("Factorial:", factorial(factorial_number))
except ValueError as error:
    print(error)

terms = int(input("Enter number of Fibonacci terms: "))
if terms < 0:
    print("Number of terms cannot be negative.")
else:
    print("Fibonacci series:", fibonacci(terms))

prime_number = int(input("Enter a number to check whether it is prime: "))
if is_prime(prime_number):
    print(prime_number, "is a prime number.")
else:
    print(prime_number, "is not a prime number.")
```

## Question 7 — File Handling
Create a text file, write student details into it, then reopen the file and display the contents. **Running this cell again overwrites** the previous `student_details.txt`.

```python
# Question 7: File Handling
name = input("Enter student name: ")
roll_number = input("Enter roll number: ")
branch = input("Enter branch: ")
marks = input("Enter marks: ")

file_name = "student_details.txt"

with open(file_name, "w", encoding="utf-8") as file:
    file.write(f"Name: {name}\n")
    file.write(f"Roll Number: {roll_number}\n")
    file.write(f"Branch: {branch}\n")
    file.write(f"Marks: {marks}\n")

print(f"Student details saved to {file_name}.")

with open(file_name, "r", encoding="utf-8") as file:
    contents = file.read()

print("\nContents of the file:")
print(contents)
```

## Question 10 — Mini Programming Problem: Student Record Management
A menu-driven student database using a Python dictionary. Records are indexed by **roll number** and store each student's name and marks. The menu supports add, update marks, search, delete, and display all records. Choose **6** to exit.

```python
# Question 10: Menu-driven Student Record Management
student_records = {}

while True:
    print("\n========== STUDENT RECORD MENU ==========")
    print("1. Add a student")
    print("2. Update marks")
    print("3. Search for a student")
    print("4. Delete a student")
    print("5. Display all student records")
    print("6. Exit")

    choice = input("Enter your choice (1-6): ").strip()

    if choice == "1":
        roll = input("Enter roll number: ").strip()
        if not roll:
            print("Roll number cannot be empty.")
        elif roll in student_records:
            print("A student with this roll number already exists.")
        else:
            name = input("Enter student name: ").strip()
            if not name:
                print("Student name cannot be empty.")
                continue
            try:
                marks = float(input("Enter marks: "))
            except ValueError:
                print("Invalid marks. Student not added.")
                continue
            student_records[roll] = {"name": name, "marks": marks}
            print("Student added successfully.")

    elif choice == "2":
        roll = input("Enter roll number to update: ").strip()
        if roll in student_records:
            try:
                new_marks = float(input("Enter new marks: "))
                student_records[roll]["marks"] = new_marks
                print("Marks updated successfully.")
            except ValueError:
                print("Invalid marks. No changes made.")
        else:
            print("Student not found.")

    elif choice == "3":
        roll = input("Enter roll number to search: ").strip()
        if roll in student_records:
            student = student_records[roll]
            print("Roll Number:", roll)
            print("Name:", student["name"])
            print("Marks:", student["marks"])
        else:
            print("Student not found.")

    elif choice == "4":
        roll = input("Enter roll number to delete: ").strip()
        if roll in student_records:
            del student_records[roll]
            print("Student deleted successfully.")
        else:
            print("Student not found.")

    elif choice == "5":
        if not student_records:
            print("No student records available.")
        else:
            print("\nAll Student Records:")
            for roll, student in student_records.items():
                print(f"Roll: {roll} | Name: {student['name']} | Marks: {student['marks']}")

    elif choice == "6":
        print("Exiting Student Record Management.")
        break

    else:
        print("Invalid choice. Please enter a number from 1 to 6.")
```
